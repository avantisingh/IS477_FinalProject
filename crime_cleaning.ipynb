```python
import pandas as pd

crime_df = pd.read_csv("crime.csv")

crime_df.head()
```

```python
crime_df["Date"] = pd.to_datetime(crime_df["Date"], errors="coerce")

crime_df = crime_df[
    (crime_df["Date"].dt.year >= 2016) &
    (crime_df["Date"].dt.year <= 2023)
]

crime_df["Year"] = crime_df["Date"].dt.year
```

```python
crime_df = crime_df[[
    "Date",
    "Block",
    "Description",
    "Primary Type",
    "Community Area",
    "Year",
    "Arrest",
    "Latitude",
    "Longitude"
]]
```
